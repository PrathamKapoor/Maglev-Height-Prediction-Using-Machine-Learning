# 05 — Physics-Inspired Features

What we CAN derive (verified variables):
- `time_squared` (nonlinear time dynamics)
- `control_squared` (nonlinear PWM effect)
- `time_x_control` (interaction)
- `control_deviation_from_mean` (relative intensity)

What we CANNOT derive (not in dataset, not invented):
- `current_squared` (no current measurement)
- `inverse_gap` (no gap measurement)
- `position_reference_deviation` (would use target — leakage)

Verified results: Adding physics-inspired features does not significantly change MAE (remains near 0) because control and position are almost perfectly correlated in this closed-loop dataset. This is documented, not hidden.
