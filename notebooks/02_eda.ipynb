# 02 — EDA (Verified Results)

Only verified statistics from actual dataset inspection.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('..') / 'src'))
from dataset import MagLevDataset
loader = MagLevDataset()
import numpy as np
arr = loader.load('Classic_PID.mat')
print('Shape:', arr.shape)
print('NaN count:', np.sum(np.isnan(arr)))
print('Duplicate rows:', arr.shape[0] - np.unique(arr, axis=0).shape[0])
print('Correlation control-position:', np.corrcoef(arr[:,2], arr[:,1])[0,1])
```

Note: Correlation between control and position is ≈ 1.0 (verified by `np.corrcoef`). This reflects the Arduino control loop computing PWM directly from position error (`x - xref`). This is a physical reality of the dataset, not a modeling error.
