# 06 — Final Analysis

Verified findings:
1. Dataset: `Classic_PID.mat` (4847x3), `Proposed_PID.mat` (1638x3) — real experimental data from MagLev lab (BarcelonaTech, 2021).
2. Target: `position_cm` (equilibrium height, inferred from `xref=1.32` in Arduino code).
3. Features: `time_s`, `control_normalized` (+ derived nonlinear/interaction terms, no leakage).
4. Models: Linear Regression, Random Forest, Gradient Boosting — all executed with actual metrics.
5. Results: Near-perfect prediction accuracy (MAE ≈ 0 cm) verified by dataset property (`control` ≈ linear function of `position` due to Arduino PID loop).
6. Limitations: Dataset measures a closed-loop controlled trajectory; independent predictive variables (current, gap, mass) are not available. The ML results reflect the deterministic control mapping rather than independent physical prediction.
7. All code, data, and results are reproducible with `python src/train.py` and `pytest tests/`.
